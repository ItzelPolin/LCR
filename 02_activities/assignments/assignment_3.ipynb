```python
# Import standard libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

#### **Question 2:** 
#### Data-visualization

Let's create plots to visualize the relationships between the features (e.g., alcohol, malic acid, ash...).


```python
def plot_feature_pairs(data, feature_names, color_labels=None, title_prefix=''):
    """
    Helper function to create scatter plots for all possible pairs of features.
    
    Parameters:
    - data: DataFrame containing the features to be plotted.
    - feature_names: List of feature names to be used in plotting.
    - color_labels: Optional. Cluster or class labels to color the scatter plots.
    - title_prefix: Optional. Prefix for plot titles to distinguish between different sets of plots.
    """
    # Create a figure for the scatter plots
    plt.figure(figsize=(60, 60))
    
    # Counter for subplot index
    plot_number = 1
    
    # Loop through each pair of features
    for i in range(len(feature_names)):
        for j in range(i + 1, len(feature_names)):
            plt.subplot(len(feature_names)-1, len(feature_names)-1, plot_number)
            
            # Scatter plot colored by labels if provided
            if color_labels is not None:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], 
                            c=color_labels, cmap='viridis', alpha=0.7)
            else:
                plt.scatter(data[feature_names[i]], data[feature_names[j]], alpha=0.7)
            
            plt.xlabel(feature_names[i])
            plt.ylabel(feature_names[j])
            plt.title(f'{title_prefix}{feature_names[i]} vs {feature_names[j]}')
            
            # Increment the plot number
            plot_number += 1

    # Adjust layout to prevent overlap
    plt.tight_layout()

    # Show the plot
    plt.suptitle(f'{title_prefix}Feature Relationships', fontsize=24)
    plt.show()

# Get feature names
feature_names = wine_df.columns

# Use the helper function to plot scatter plots without coloring by cluster labels
plot_feature_pairs(wine_df, feature_names, title_prefix='Original Data: ')
```

**Question:**

Additionally, color intensity and hue show an inverse relationship, suggesting different pigment and aging characteristics across wine groups. These structured patterns imply that the dataset contains naturally separable groups, making it well-suited for clustering algorithms like KMeans to distinguish between different wine species (classes) based on their chemical profiles.


#### **Question 3:** 
#### Data cleaning

```python
# Initialize the StandardScaler
scaler = StandardScaler()

# Scale all the features in the dataset
scaled_features = scaler.fit_transform(wine_df)

# Create a new DataFrame with scaled features
scaled_wine_df = pd.DataFrame(scaled_features, columns=wine_data.feature_names)

# Display the first few rows of the scaled DataFrame
print(scaled_wine_df.head())
```

Why is it important to standardize the features of a dataset before applying clustering algorithms like K-Means? Discuss the implications of using unstandardized data in your analysis.

It is important to standardize features before applying K-Means because the algorithm is distance-based and uses Euclidean distance to form clusters. If the data is not standardized, features with larger numeric ranges (such as proline or magnesium) will dominate the distance calculations, causing the clustering to be biased toward those variables. This can lead to misleading clusters that reflect scale differences rather than true similarities in the data. Standardization ensures that all features contribute equally to the clustering process and improves the accuracy and interpretability of the results.> Your answer here ... 

#### **Question 4:** 
#### K-means clustering 

Apply the K-Means clustering algorithm to the Wine dataset. Choose the value 3 for the number of clusters (`k=3`) and fit the model. Assign cluster labels to the original data and add them as a new column in the DataFrame.

```python
# Your answer...

kmeans = KMeans(n_clusters=3, random_state=0, n_init="auto").fit(scaled_wine_df)
clustered_wine_data = wine_df.assign(Cluster=kmeans.labels_)


# Use the helper function to plot scatter plots, colored by cluster labels
plot_feature_pairs(clustered_wine_data, feature_names, color_labels=clustered_wine_data['Cluster'], title_prefix='Clustered Data: ')
```

We chose `k=3` for the number of clusters arbitrarily. However, in a real-world scenario, it is important to determine the optimal number of clusters using appropriate methods.

**Question**: What is one method commonly used to determine the optimal number of clusters in K-means clustering, and why is this method helpful?

Your answer here...
The Elbow Method. This method plots the within-cluster sum of squares (WCSS) against different values of k and identifies the point where the rate of decrease sharply slows down, forming an “elbow.” This point represents a balance between minimizing cluster variance and avoiding unnecessary additional clusters, helping to select a value of k that best captures the natural structure of the data. 

#### **Question 5:** 
#### Bootstrapping 

Implement bootstrapping on the mean of `color intensity`. Generate 10000 bootstrap samples, calculate the mean for each sample, and compute a 90% confidence interval.

```python
# Your answer here...

mean_color_intensity = wine_df['color_intensity'].mean()

np.random.seed(123)

lower_bound = np.percentile([wine_df['color_intensity'].sample(frac=1, replace=True).mean() for _ in range(10000)], 5)
upper_bound = np.percentile([wine_df['color_intensity'].sample(frac=1, replace=True).mean() for _ in range(10000)], 95)

# Display the result
print(f"Mean of Color Intensity: {mean_color_intensity}")
print(f"90% Confidence Interval of Mean Color Intensity: ({lower_bound}, {upper_bound})")
```

**Question:**
- Why do we use bootstrapping in this context? What does it help us understand about the mean?

We use bootstrapping to see how much the mean might change if we took many different samples from the same dataset. By repeatedly resampling the data, we can measure how consistent the average color intensity is and create a confidence interval. This helps us understand how reliable our calculated mean is and how much uncertainty there is in our estimate

**Question:**
- What is the purpose of calculating the confidence interval from the bootstrap samples? How does it help us interpret the variability of the estimate?

The confidence interval shows the range of values where the true mean is likely to fall. It helps us understand how much the mean could vary if we collected different samples. A narrow interval means the estimate is very stable and reliable, while a wider interval means there is more uncertainty in the estimate.

**Question:**

- Reflect on the variability observed in the bootstrapped means and discuss whether the mean of the color intensity appears to be a stable and reliable estimate based on the confidence interval and the spread of the bootstrapped means.

The bootstrapped means show relatively little spread, and the 90% confidence interval for the mean color intensity is fairly narrow (approximately 4.78 to 5.35). This indicates that the estimated mean does not change much across repeated resamples of the data. Because the interval is tight and centered around the observed mean (about 5.06), the mean color intensity appears to be a stable and reliable estimate. This suggests that the sample provides a consistent representation of the average color intensity in the dataset.
